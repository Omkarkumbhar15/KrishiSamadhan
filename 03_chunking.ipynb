```python
# ================================
# IMPORTS
# ================================
import json
import os


# ================================
# PATHS
# ================================
input_path = "data/processed/cleaned_data.json"
output_path = "data/processed/chunks.json"


# ================================
# LOAD DATA
# ================================
if not os.path.exists(input_path):
    raise FileNotFoundError(f"{input_path} not found. Run preprocessing notebook first.")

with open(input_path, "r", encoding="utf-8") as f:
    data = json.load(f)

print(f"Loaded {len(data)} cleaned documents")


# ================================
# CHUNKING FUNCTION
# ================================

def chunk_text(text, chunk_size=500, overlap=100):
    words = text.split()
    chunks = []

    start = 0
    text_length = len(words)

    while start < text_length:
        end = start + chunk_size
        chunk_words = words[start:end]

        chunk_text = " ".join(chunk_words)

        if len(chunk_text.strip()) > 0:
            chunks.append(chunk_text)

        # Move start with overlap
        start += (chunk_size - overlap)

    return chunks


# ================================
# APPLY CHUNKING
# ================================
all_chunks = []
chunk_id = 0

print("\n🔹 Creating chunks...\n")

for doc in data:
    text = doc.get("text", "")
    
    if not text:
        continue

    chunks = chunk_text(text, chunk_size=500, overlap=100)

    for chunk in chunks:
        all_chunks.append({
            "chunk_id": chunk_id,
            "text": chunk,
            "source": doc.get("source"),
            "type": doc.get("type"),
            "domain": doc.get("domain")
        })
        chunk_id += 1


print(f"Total chunks created: {len(all_chunks)}")


# ================================
# SAVE CHUNKS
# ================================
with open(output_path, "w", encoding="utf-8") as f:
    json.dump(all_chunks, f, indent=4)

print(f"\n✅ Chunks saved to {output_path}")


# ================================
# PREVIEW
# ================================
print("\n🔹 Sample Chunks:\n")

for item in all_chunks[:2]:
    print(item)
    print("\n----------------------\n")
```
